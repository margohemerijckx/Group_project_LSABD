```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA

# 1. Load the data and verify that the sample columns in the count matrix
# exactly match the sample descriptions provided in the metadata.
meta = pd.read_csv("GSE80655_metadata.csv")
counts = pd.read_csv("GSE80655_counts.csv", index_col=0)

assert list(counts.columns) == list(meta["description"])

#2. Changing the data types where needed
# donor_id: from integer to string
# sex: from string to integer
print(meta.dtypes)
meta["donor_id"] = meta["donor_id"].astype(str)
meta["sex"] = meta["sex"].map({
    "F": 0,
    "M": 1
})

# 3. Impute missing brain_pH values for donor 4463 using the overall mean.
meta["brain_pH"] = meta["brain_pH"].fillna(meta["brain_pH"].mean())

# 4. Explore the study design and sample composition.
# Most of the 96 donors were sampled across multiple brain regions,
# meaning that the 281 observations are not independent. In addition,
# the sex distribution is imbalanced and the majority of donors are Caucasian.
print(pd.crosstab(meta["brain_region"], meta["diagnosis"]))
print("Unique donors:", meta["donor_id"].nunique())
print("Samples per donor:\n", meta.groupby("donor_id").size().value_counts())
print("Sex distribution (donor level):\n",
      meta.drop_duplicates("donor_id")["sex"].value_counts())
# 5. Filter lowly expressed genes to reduce technical noise and improve
# the reliability of downstream normalization and statistical analyses.
# Retain genes with at least 10 counts in a minimum of 22 samples.
keep_genes = (counts >= 10).sum(axis=1) >= 22
counts_filt = counts.loc[keep_genes].copy()

print("Number of genes after filtering:", counts_filt.shape[0])

# Visualize library size distribution after gene filtering.
# Substantial differences in library size indicate the need for
# normalization prior to downstream analyses.

lib_size_M = counts_filt.sum(axis=0) / 1e6

plt.figure(figsize=(8, 5))

sns.histplot(
    lib_size_M,
    bins=25,
    kde=True,
    color="#2b5c8f"
)

plt.axvline(
    lib_size_M.median(),
    color="red",
    linestyle="--",
    label=f"Median: {lib_size_M.median():.1f}M"
)

plt.title("Library size distribution after gene filtering")
plt.xlabel("Total counts per sample (millions)")
plt.ylabel("Number of samples")
plt.legend()

plt.tight_layout()
plt.show()

# 6. Perform median-of-ratios normalization (DESeq2 method).
# This approach corrects for differences in sequencing depth across samples
# by estimating sample-specific size factors from the ratio of each gene's
# count to its geometric mean across all samples. Normalized counts are
# subsequently transformed to the log2 scale for downstream analyses.
log_counts = np.log(counts_filt.replace(0, np.nan))
geom_means = np.exp(log_counts.mean(axis=1))

size_factors = counts_filt.div(geom_means, axis=0).median(axis=0)
norm_counts = counts_filt.div(size_factors, axis=1)

log_norm = np.log2(norm_counts + 1)


# -------------------------------------------------------------
# Visualize expression distributions before and after normalization
# The first plot shows the raw expression values, whereas the second
# plot shows the normalized expression values after median-of-ratios
# normalization.
# -------------------------------------------------------------

log_raw = np.log2(counts_filt + 1)

# Compare expression distributions before and after normalization
# for the first 25 samples.

sub_cols = counts_filt.columns[:25]

plt.figure(figsize=(14, 5))

bp1 = plt.boxplot(
    [log_raw[c] for c in sub_cols],
    positions=np.arange(25) - 0.18,
    widths=0.32,
    patch_artist=True,
    showfliers=False
)

bp2 = plt.boxplot(
    [log_norm[c] for c in sub_cols],
    positions=np.arange(25) + 0.18,
    widths=0.32,
    patch_artist=True,
    showfliers=False
)

for p in bp1["boxes"]:
    p.set_facecolor("#b0c4de")  # light blue = before normalization

for p in bp2["boxes"]:
    p.set_facecolor("#2b5c8f")  # dark blue = after normalization

plt.title("Expression distributions before (light) and after (dark) normalization")
plt.xlabel("First 25 samples")
plt.ylabel("log2(counts + 1)")
plt.xticks([])

from matplotlib.patches import Patch
plt.legend(
    handles=[
        Patch(facecolor="#b0c4de", label="Before normalization"),
        Patch(facecolor="#2b5c8f", label="After normalization")
    ],
    loc="upper right"
)

plt.tight_layout()
plt.show()


# 7. Perform principal component analysis (PCA) on the normalized
# expression matrix to summarize major sources of variation.
from sklearn.decomposition import PCA
from scipy.cluster.hierarchy import linkage, dendrogram

pca = PCA(n_components=5, random_state=42)
pcs = pca.fit_transform(log_norm.T)

pca_df = pd.DataFrame(
    pcs,
    columns=[f"PC{i+1}" for i in range(5)],
    index=counts_filt.columns
)

pca_df = pca_df.join(meta.set_index("description"))

var_exp = pca.explained_variance_ratio_ * 100

plt.figure(figsize=(7, 6))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="brain_region",
    style="diagnosis",
    s=55,
    alpha=0.85
)

plt.title("PCA of normalized expression data")
plt.xlabel(f"PC1 ({var_exp[0]:.1f}% variance explained)")
plt.ylabel(f"PC2 ({var_exp[1]:.1f}% variance explained)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(7, 6))

sc = plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    c=pca_df["brain_pH"],
    cmap="viridis",
    s=45,
    alpha=0.85
)

imp = pca_df[pca_df["donor_id"] == "4463"]

plt.scatter(
    imp["PC1"],
    imp["PC2"],
    facecolors="none",
    edgecolors="red",
    s=140,
    linewidths=1.8,
    label="Donor 4463 (mean pH)"
)

plt.colorbar(sc, label="brain_pH")

plt.title("PCA coloured by brain pH")
plt.xlabel(f"PC1 ({var_exp[0]:.1f}% variance explained)")
plt.ylabel(f"PC2 ({var_exp[1]:.1f}% variance explained)")
plt.legend()
plt.tight_layout()
plt.show()

# Select the 500 most variable genes and perform hierarchical clustering
# using Ward's linkage and Euclidean distance.
top500 = log_norm.var(axis=1).nlargest(500).index
Z = linkage(
    log_norm.loc[top500].T,
    method="ward",
    metric="euclidean"
)

plt.figure(figsize=(14, 4.5))

dendrogram(
    Z,
    labels=pca_df["brain_region"].values,
    leaf_rotation=90,
    leaf_font_size=6,
    color_threshold=0
)

plt.title("Hierarchical clustering (Ward linkage, 500 most variable genes)")
plt.ylabel("Euclidean distance")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf
from scipy import stats

# 1. Extract EGR1 (ENSG00000120738) expression values from the normalized
# expression matrix and prepare the metadata for statistical modeling.
egr1_id = "ENSG00000120738"
assert egr1_id in log_norm.index

df = meta.copy()
df["egr1_log_norm"] = log_norm.loc[egr1_id, df["description"]].values
df["log_lib_size"] = np.log10(counts_filt.sum(axis=0).loc[df["description"]].values)

df["diagnosis"] = pd.Categorical(
    df["diagnosis"],
    categories=["Control", "Schizophrenia", "Bipolar Disorder", "Major Depression"]
)
df["brain_region"] = pd.Categorical(
    df["brain_region"],
    categories=["AnCg", "DLPFC", "nAcc"]
)
df["sex"] = pd.Categorical(df["sex"])
df["donor_id"] = df["donor_id"].astype(str)

for col in ["age", "PMI", "brain_pH", "log_lib_size"]:
    df[col + "_z"] = (df[col] - df[col].mean()) / df[col].std()

# 2. Fit mixed-effects models with donor ID included as a random effect
# to account for repeated measurements obtained from the same donor.
formula_full = "egr1_log_norm ~ age_z + sex + PMI_z + brain_pH_z + log_lib_size_z + brain_region * diagnosis"
formula_add  = "egr1_log_norm ~ age_z + sex + PMI_z + brain_pH_z + log_lib_size_z + brain_region + diagnosis"

# Fit the full and additive mixed-effects models using Maximum
# Likelihood (ML) estimation. ML is required for a valid
# likelihood-ratio test (LRT) comparing nested models.
m_full_ml = smf.mixedlm(formula_full, df, groups=df["donor_id"]).fit(reml=False)
m_add_ml  = smf.mixedlm(formula_add,  df, groups=df["donor_id"]).fit(reml=False)

lr_stat = 2 * (m_full_ml.llf - m_add_ml.llf)
df_diff = len(m_full_ml.fe_params) - len(m_add_ml.fe_params)
p_lrt = stats.chi2.sf(lr_stat, df=df_diff)
print(f"LRT interactie: chi2({df_diff}) = {lr_stat:.3f}, p = {p_lrt:.4e}")

# Fit the final mixed-effects model using restricted maximum likelihood (REML).
m_full = smf.mixedlm(formula_full, df, groups=df["donor_id"]).fit(reml=True)
print(m_full.summary().tables[1])

# 3. Perform post-hoc contrasts to compare schizophrenia and control
# samples within each brain region.
fe = m_full.fe_params
cov_fe = m_full.cov_params().loc[fe.index, fe.index]
idx_sz = list(fe.index).index("diagnosis[T.Schizophrenia]")
idx_dlpfc_sz = list(fe.index).index("brain_region[T.DLPFC]:diagnosis[T.Schizophrenia]")
idx_nacc_sz = list(fe.index).index("brain_region[T.nAcc]:diagnosis[T.Schizophrenia]")

for label, extra_idx in [("AnCg", None), ("DLPFC", idx_dlpfc_sz), ("nAcc", idx_nacc_sz)]:
    vec = np.zeros(len(fe))
    vec[idx_sz] = 1.0
    if extra_idx is not None:
        vec[extra_idx] = 1.0
    est = float(np.dot(vec, fe))
    se = float(np.sqrt(vec @ cov_fe.values @ vec))
    ci_low, ci_high = est - 1.96 * se, est + 1.96 * se
    p_val = 2 * stats.norm.sf(abs(est / se))
    print(f"{label} (SZ vs CTL): log2_diff = {est:.3f} [95% BI {ci_low:.3f}, {ci_high:.3f}], p = {p_val:.4e}")

# 4. Visualize EGR1 expression patterns and assess model assumptions
# using residual diagnostic boxplots.
df["fitted"] = m_full.fittedvalues
df["resid"] = m_full.resid

fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))
order_diag = ["Control", "Schizophrenia", "Bipolar Disorder", "Major Depression"]
palette_diag = {
    "Control": "#6c757d",
    "Schizophrenia": "#d62728",
    "Bipolar Disorder": "#1f77b4",
    "Major Depression": "#2ca02c"
}

sns.boxplot(data=df, x="brain_region", y="egr1_log_norm", hue="diagnosis",
            hue_order=order_diag, palette=palette_diag, ax=axes[0], fliersize=2.5)
axes[0].set_title("EGR1 (ENSG00000120738) per regio & diagnose")
axes[0].set_xlabel("Hersenregio")
axes[0].set_ylabel("log2(genormaliseerde counts + 1)")
axes[0].legend(title="Diagnose", fontsize=7.5, title_fontsize=8, loc="upper right")

axes[1].scatter(df["fitted"], df["resid"], alpha=0.65, s=25, color="#2b5c8f")
axes[1].axhline(0, color="red", linestyle="--", linewidth=1)
axes[1].set_title("Residuen vs. voorspelde waarden")
axes[1].set_xlabel("Voorspelde waarde (fitted)")
axes[1].set_ylabel("Residu")

stats.probplot(df["resid"], dist="norm", plot=axes[2])
axes[2].set_title("QQ-plot van residuen")

plt.tight_layout()
plt.show()
```
