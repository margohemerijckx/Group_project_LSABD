```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA

# 1. Data laden en controleren of de samplekolommen in de countmatrix exact overeenkomen met description in de metadata 
meta = pd.read_csv("GSE80655_metadata.csv")
counts = pd.read_csv("GSE80655_counts.csv", index_col=0)
assert list(counts.columns) == list(meta["description"])

# 2. Imputatie van ontbrekende brain_pH bij donor_id 4463 met het gemiddelde (mean)
mean_ph = meta["brain_pH"].mean()  # 6.901475
meta["brain_pH"] = meta["brain_pH"].fillna(mean_ph)

# 3. Omdat de meeste van de 96 donoren in meerdere regio's zijn bemonsterd zijn de 281 metingen niet onafhankelijk. Daarnaast is sex scheef verdeeld en ethnicity grotendeels caucasian.
print(pd.crosstab(meta["brain_region"], meta["diagnosis"]))
print("Unieke donoren:", meta["donor_id"].nunique())
print("Samples per donor:\n", meta.groupby("donor_id").size().value_counts())
print("Geslacht (op donorniveau):\n", meta.drop_duplicates("donor_id")["sex"].value_counts())

# 4. Genfiltering (genen verwijderen met zeer lage of nul-tellingen zodat ruis en extreem lage counts de normalisatei en latere testen niet verstoren. Gebruiken dremple minstens 10 counts in minstens 22 samples)
keep_genes = (counts >= 10).sum(axis=1) >= 22
counts_filt = counts.loc[keep_genes].copy()
print("Aantal genen na filtering:", counts_filt.shape[0])

# 5. Median-of-ratios normalisatie (DESeq2-methode), corrigeren voor verschillen in sequencing diepte tss samples door per sample een size factor te schatten obv de ratio tov het geometrische gemiddelde per gen en daarna genormaliseerde tellingen omzetten in log2 schaal.
log_counts = np.log(counts_filt.replace(0, np.nan))
geom_means = np.exp(log_counts.mean(axis=1))
size_factors = counts_filt.div(geom_means, axis=0).median(axis=0)
norm_counts = counts_filt.div(size_factors, axis=1)
log_norm = np.log2(norm_counts + 1)

# 6. PCA op genormaliseerde matrix
from sklearn.decomposition import PCA
from scipy.cluster.hierarchy import linkage, dendrogram

pca = PCA(n_components=5, random_state=42)
pcs = pca.fit_transform(log_norm.T)
pca_df = pd.DataFrame(pcs, columns=[f"PC{i+1}" for i in range(5)], index=counts_filt.columns)
pca_df = pca_df.join(meta.set_index("description"))

top500 = log_norm.var(axis=1).nlargest(500).index
Z = linkage(log_norm.loc[top500].T, method="ward", metric="euclidean")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.cluster.hierarchy import linkage, dendrogram

var_exp = pca.explained_variance_ratio_ * 100
pca_df["lib_size_M"] = counts_filt.sum(axis=0).values / 1e6
log_raw = np.log2(counts_filt + 1)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Library size verdeling
sns.histplot(pca_df["lib_size_M"], bins=25, kde=True, color="#2b5c8f", ax=axes[0, 0])
axes[0, 0].axvline(pca_df["lib_size_M"].median(), color="red", linestyle="--",
                   label=f"Mediaan: {pca_df['lib_size_M'].median():.1f}M")
axes[0, 0].set_title("1. Library size verdeling (na genfiltering)")
axes[0, 0].set_xlabel("Totaal aantal counts per sample (miljoenen)")
axes[0, 0].set_ylabel("Aantal samples")
axes[0, 0].legend()

# 2. Expressieverdeling voor vs na normalisatie (eerste 25 samples)
sub_cols = counts_filt.columns[:25]
bp1 = axes[0, 1].boxplot([log_raw[c] for c in sub_cols], positions=np.arange(25)-0.18,
                         widths=0.32, patch_artist=True, showfliers=False)
bp2 = axes[0, 1].boxplot([log_norm[c] for c in sub_cols], positions=np.arange(25)+0.18,
                         widths=0.32, patch_artist=True, showfliers=False)
for p in bp1["boxes"]: p.set_facecolor("#b0c4de")
for p in bp2["boxes"]: p.set_facecolor("#2b5c8f")
axes[0, 1].set_title("2. Expressieverdeling vóór (licht) vs. ná (donker) normalisatie")
axes[0, 1].set_ylabel("log2(counts + 1)")
axes[0, 1].set_xlabel("Eerste 25 samples")
axes[0, 1].set_xticks([])

# 3. PCA op hersenregio en diagnose
sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="brain_region", style="diagnosis",
                s=55, alpha=0.85, ax=axes[1, 0])
axes[1, 0].set_title("3. PCA op genormaliseerde data (regio & diagnose)")
axes[1, 0].set_xlabel(f"PC1 ({var_exp[0]:.1f}% variantie)")
axes[1, 0].set_ylabel(f"PC2 ({var_exp[1]:.1f}% variantie)")

# 4. PCA op brain_pH (incl. geïmputeerde donor 4463)
sc = axes[1, 1].scatter(pca_df["PC1"], pca_df["PC2"], c=pca_df["brain_pH"], cmap="viridis", s=45, alpha=0.85)
imp = pca_df[pca_df["donor_id"] == 4463]
axes[1, 1].scatter(imp["PC1"], imp["PC2"], facecolors="none", edgecolors="red", s=140,
                   linewidths=1.8, label="Donor 4463 (mean pH)")
plt.colorbar(sc, ax=axes[1, 1], label="brain_pH")
axes[1, 1].set_title("4. PCA gekleurd op brain_pH (QC- & outliercheck)")
axes[1, 1].set_xlabel(f"PC1 ({var_exp[0]:.1f}% variantie)")
axes[1, 1].set_ylabel(f"PC2 ({var_exp[1]:.1f}% variantie)")
axes[1, 1].legend(loc="best")

plt.tight_layout()
plt.show()

# 5. Hiërarchische clustering (Ward, Euclidische afstand op top 500 variabele genen)
top500_genes = log_norm.var(axis=1).nlargest(500).index
Z = linkage(log_norm.loc[top500_genes].T, method="ward", metric="euclidean")

plt.figure(figsize=(14, 4.5))
dendrogram(Z, labels=pca_df["brain_region"].values, leaf_rotation=90, leaf_font_size=6, color_threshold=0)
plt.title("5. Hiërarchische clustering (Ward, top 500 variabele genen)")
plt.ylabel("Euclidische afstand")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf
from scipy import stats

# 1. EGR1 (ENSG00000120738) uit genormaliseerde matrix halen
egr1_id = "ENSG00000120738"
assert egr1_id in log_norm.index

df = meta.copy()
df["egr1_log_norm"] = log_norm.loc[egr1_id, df["description"]].values
df["log_lib_size"] = np.log10(counts_filt.sum(axis=0).loc[df["description"]].values)

df["diagnosis"] = pd.Categorical(
    df["diagnosis"],
    categories=["Control", "Schizophrenia", "Bipolar Disorder", "Major Depression"]
)
df["brain_region"] = pd.Categorical(
    df["brain_region"],
    categories=["AnCg", "DLPFC", "nAcc"]
)
df["sex"] = pd.Categorical(df["sex"])
df["donor_id"] = df["donor_id"].astype(str)

for col in ["age", "PMI", "brain_pH", "log_lib_size"]:
    df[col + "_z"] = (df[col] - df[col].mean()) / df[col].std()

# 2. Mixed-effects model met donor_id als random effect
formula_full = "egr1_log_norm ~ age_z + sex + PMI_z + brain_pH_z + log_lib_size_z + brain_region * diagnosis"
formula_add  = "egr1_log_norm ~ age_z + sex + PMI_z + brain_pH_z + log_lib_size_z + brain_region + diagnosis"

# LRT voor interactie (met ML)
m_full_ml = smf.mixedlm(formula_full, df, groups=df["donor_id"]).fit(reml=False)
m_add_ml  = smf.mixedlm(formula_add,  df, groups=df["donor_id"]).fit(reml=False)

lr_stat = 2 * (m_full_ml.llf - m_add_ml.llf)
df_diff = len(m_full_ml.fe_params) - len(m_add_ml.fe_params)
p_lrt = stats.chi2.sf(lr_stat, df=df_diff)
print(f"LRT interactie: chi2({df_diff}) = {lr_stat:.3f}, p = {p_lrt:.4e}")

# Finale model (met REML)
m_full = smf.mixedlm(formula_full, df, groups=df["donor_id"]).fit(reml=True)
print(m_full.summary().tables[1])

# 3. Post-hoc contrasten: Schizophrenia vs Control per hersenregio
fe = m_full.fe_params
cov_fe = m_full.cov_params().loc[fe.index, fe.index]
idx_sz = list(fe.index).index("diagnosis[T.Schizophrenia]")
idx_dlpfc_sz = list(fe.index).index("brain_region[T.DLPFC]:diagnosis[T.Schizophrenia]")
idx_nacc_sz = list(fe.index).index("brain_region[T.nAcc]:diagnosis[T.Schizophrenia]")

for label, extra_idx in [("AnCg", None), ("DLPFC", idx_dlpfc_sz), ("nAcc", idx_nacc_sz)]:
    vec = np.zeros(len(fe))
    vec[idx_sz] = 1.0
    if extra_idx is not None:
        vec[extra_idx] = 1.0
    est = float(np.dot(vec, fe))
    se = float(np.sqrt(vec @ cov_fe.values @ vec))
    ci_low, ci_high = est - 1.96 * se, est + 1.96 * se
    p_val = 2 * stats.norm.sf(abs(est / se))
    print(f"{label} (SZ vs CTL): log2_diff = {est:.3f} [95% BI {ci_low:.3f}, {ci_high:.3f}], p = {p_val:.4e}")

# 4. Boxplot + residudiagnostiek
df["fitted"] = m_full.fittedvalues
df["resid"] = m_full.resid

fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))
order_diag = ["Control", "Schizophrenia", "Bipolar Disorder", "Major Depression"]
palette_diag = {
    "Control": "#6c757d",
    "Schizophrenia": "#d62728",
    "Bipolar Disorder": "#1f77b4",
    "Major Depression": "#2ca02c"
}

sns.boxplot(data=df, x="brain_region", y="egr1_log_norm", hue="diagnosis",
            hue_order=order_diag, palette=palette_diag, ax=axes[0], fliersize=2.5)
axes[0].set_title("EGR1 (ENSG00000120738) per regio & diagnose")
axes[0].set_xlabel("Hersenregio")
axes[0].set_ylabel("log2(genormaliseerde counts + 1)")
axes[0].legend(title="Diagnose", fontsize=7.5, title_fontsize=8, loc="upper right")

axes[1].scatter(df["fitted"], df["resid"], alpha=0.65, s=25, color="#2b5c8f")
axes[1].axhline(0, color="red", linestyle="--", linewidth=1)
axes[1].set_title("Residuen vs. voorspelde waarden")
axes[1].set_xlabel("Voorspelde waarde (fitted)")
axes[1].set_ylabel("Residu")

stats.probplot(df["resid"], dist="norm", plot=axes[2])
axes[2].set_title("QQ-plot van residuen")

plt.tight_layout()
plt.show()
```
